# Bước 1: Khám phá và Phân tích Dữ liệu (EDA)

Mục tiêu của bước này là hiểu rõ về tập dữ liệu hình ảnh chữ số viết tay (MNIST/Digit Recognizer) trước khi tiến hành tiền xử lý và đưa vào mô hình CNN.

### Kế hoạch chi tiết:
1. **Tải các thư viện cần thiết**: Cài đặt và import `pandas`, `numpy`, `matplotlib`, `seaborn`.
2. **Đọc dữ liệu**: Tải file `train.csv` và `test.csv` từ thư mục `data/digit-recognizer/`.
3. **Kiểm tra thông tin tổng quan**: Xem kích thước dữ liệu (số lượng mẫu, số lượng features), kiểm tra xem có ô dữ liệu nào bị trống (NaN/Missing values) không.
4. **Phân tích phân phối Nhãn (Label Distribution)**: Đảm bảo số lượng hình ảnh cho các chữ số từ 0 đến 9 là tương đối đồng đều, tránh hiện tượng mất cân bằng dữ liệu (imbalanced data).
5. **Trực quan hóa Dữ liệu (Data Visualization)**: Reshape (biến đổi hình dạng) các hàng pixel độ dài 784 thành các mảng 2D kích thước 28x28 để hiển thị dưới dạng hình ảnh, giúp kiểm chứng bằng mắt.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Liệt kê các file trong thư mục dữ liệu
data_path = 'data/digit-recognizer/'
print("Các file dữ liệu:", os.listdir(data_path))
```

### 1. Đọc dữ liệu
Sử dụng pandas để đọc dữ liệu từ file CSV.

```python
train_df = pd.read_csv(data_path + 'train.csv')
test_df = pd.read_csv(data_path + 'test.csv')

print(f"Kích thước tập train: {train_df.shape}")
print(f"Kích thước tập test:  {test_df.shape}")
```

### 2. Xem trước dữ liệu và Kiểm tra Missing Values
Mỗi hàng trong tập train sẽ bao gồm 1 cột `label` (nhãn) và 784 cột `pixel0` đến `pixel783` đại diện cho giá trị điểm ảnh.

```python
display(train_df.head())

# Kiểm tra giá trị thiếu (NaN)
print("\nTổng số giá trị NaN trong tập train:", train_df.isnull().sum().sum())
print("Tổng số giá trị NaN trong tập test:", test_df.isnull().sum().sum())
```

### 3. Phân phối Nhãn (Label Distribution)
Một tập dữ liệu tốt cần có sự phân bố đồng đều giữa các nhóm cần phân loại để mô hình không bị thiên lệch.

```python
plt.figure(figsize=(10, 5))
sns.countplot(x='label', data=train_df, palette='viridis')
plt.title('Phân phối các chữ số trong tập huấn luyện (Train)')
plt.xlabel('Chữ số (Label)')
plt.ylabel('Số lượng')
plt.show()

print(train_df['label'].value_counts().sort_index())
```

### 4. Trực quan hóa Dữ liệu (Hiển thị ảnh)
Dữ liệu pixel đang ở dạng mảng 1 chiều (1D) với 784 điểm ảnh. Chúng ta cần reshape chúng thành mảng 2 chiều 28x28 để có thể hiển thị dưới dạng hình ảnh xám (grayscale).

```python
# Tách Nhãn (Y) và Đặc trưng pixel (X)
X_train = train_df.drop(labels=['label'], axis=1)
Y_train = train_df['label']

# Chọn ngẫu nhiên 9 bức ảnh để hiển thị
plt.figure(figsize=(10, 10))
for i in range(9):
    plt.subplot(3, 3, i+1)
    
    # Lấy hàng thứ i, chuyển thành numpy array và reshape thành 28x28
    img_data = X_train.iloc[i].values.reshape(28, 28)
    
    plt.imshow(img_data, cmap='gray')
    plt.title(f"Nhãn: {Y_train.iloc[i]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```
